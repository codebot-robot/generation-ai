# gRPC Fine-tuning Client

This notebook demonstrates how to use the gRPC client to trigger a fine-tuning job on the server.

```python
!pip install grpcio grpcio-tools
```

```python
import grpc
import finetuning_pb2
import finetuning_pb2_grpc

def run_finetuning(host, model_id, dataset_id, dataset_text_field, max_steps):
    with grpc.insecure_channel(host) as channel:
        stub = finetuning_pb2_grpc.FinetuningServiceStub(channel)
        request = finetuning_pb2.SFTRequest(
            model_id=model_id,
            dataset_id=dataset_id,
            dataset_text_field=dataset_text_field,
            max_steps=max_steps
        )
        
        print(f"Requesting SFT for {model_id}...")
        for response in stub.StartSFT(request):
            print(f"SERVER: {response.log_entry}")

run_finetuning(
    host='localhost:50051', 
    model_id='facebook/opt-125m', 
    dataset_id='timdettmers/openassistant-guanaco', 
    dataset_text_field='text', 
    max_steps=5
)
```
